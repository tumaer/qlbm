```python
import numpy as np
from QLBM_functions import create_QLBM_circuit, run_circuit
from utility_functions import generate_1D_initial_density

import matplotlib.pyplot as plt
```

```python
N_points_x = 32
advection_velocity = np.ones(N_points_x)*0.1
#change time steps to the desired value
timesteps = 50

shots_array = [10**5,10**6,10**7]
shots_array = [10**7]
density_result = np.zeros([len(shots_array),N_points_x])
initial_density = generate_1D_initial_density(N_POINTS_X=N_points_x, ambient_density=0.1)
qc, normalization_constant = create_QLBM_circuit(initial_density, advection_velocity, timesteps)

for ii, shots in enumerate(shots_array):
    density_result[ii,:] = run_circuit(qc=qc,shots=shots,normalization_constant=normalization_constant, recreate_paper_results=True)
```

```python
N_DISCRETE_VELOCITIES = 3
LATTICE_VELOCITIES = np.array([ 0 , 1,  -1,])
LATTICE_INDICES = np.array([1, 2,])
LATTICE_WEIGHTS = np.array([2/3, 1/6,  1/6,])
t=timesteps
digital_solution=np.copy(initial_density)

c_s = 1/np.sqrt(3)
u=advection_velocity

for ii in range(t):
    f0=LATTICE_WEIGHTS[0]*digital_solution
    f1=LATTICE_WEIGHTS[1]*digital_solution*(1+(LATTICE_VELOCITIES[1]*u)/(c_s**2))
    f2=LATTICE_WEIGHTS[2]*digital_solution*(1+(LATTICE_VELOCITIES[2]*u)/(c_s**2))
    f1=np.roll(f1,LATTICE_VELOCITIES[1],axis=0,)
    f2=np.roll(f2,LATTICE_VELOCITIES[2],axis=0,)
    digital_solution=f0+f1+f2
```

```python
plt.plot(density_result[-1,:])
plt.plot(digital_solution,'--')
plt.grid()
plt.xlim([0,N_points_x-1])
plt.xticks(np.arange(0, N_points_x, step=4))
plt.yticks(np.arange(0.09, 0.21, step=0.02))
plt.ylim([0.09,0.2])
plt.ylabel(r'$\rho$')
plt.xlabel(r'x')
plt.tight_layout()
plt.draw()
```
