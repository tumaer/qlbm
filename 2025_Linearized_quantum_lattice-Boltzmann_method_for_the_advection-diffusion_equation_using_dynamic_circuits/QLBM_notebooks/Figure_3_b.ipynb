```python
import numpy as np
from QLBM_functions import create_QLBM_circuit, run_circuit
from utility_functions import generate_2D_initial_density

import matplotlib.pyplot as plt
```

```python
N_points_x = 16
N_points_y = 16
advection_velocity = np.ones([N_points_x,N_points_y,2])*0.1
timesteps = 1

shots_array = [10000,50000,100000,500000,1000000,5000000,10000000,50000000,100000000]

density_result = np.zeros([len(shots_array),N_points_x, N_points_y])
initial_density = generate_2D_initial_density(N_POINTS_X=N_points_x, ambient_density=0.1)
qc, normalization_constant = create_QLBM_circuit(initial_density, advection_velocity, timesteps)

for ii, shots in enumerate(shots_array):
    temp = run_circuit(qc=qc,shots=shots,normalization_constant=normalization_constant, recreate_paper_results=True)
    density_result[ii,...] = temp.reshape([N_points_x,N_points_y])
```
