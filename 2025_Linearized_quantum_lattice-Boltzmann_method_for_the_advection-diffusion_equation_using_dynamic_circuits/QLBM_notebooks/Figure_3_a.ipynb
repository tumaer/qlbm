```python
import numpy as np
from QLBM_functions import create_QLBM_circuit, run_circuit
from utility_functions import generate_1D_initial_density

import matplotlib.pyplot as plt
```

```python
N_points_x = 32
advection_velocity = np.ones(N_points_x)*0.1
timesteps = 1

shots_array = [1000,5000,10000,20000,30000,40000,50000,60000,70000,80000,90000,100000,200000,300000,400000,500000,600000,700000,800000,900000,1000000,2000000,3000000,4000000,5000000,10000000,20000000,30000000,40000000,50000000,100000000]
density_result = np.zeros([len(shots_array),N_points_x])
initial_density = generate_1D_initial_density(N_POINTS_X=N_points_x, ambient_density=0.1)
qc, normalization_constant = create_QLBM_circuit(initial_density, advection_velocity, timesteps)

for ii, shots in enumerate(shots_array):
    density_result[ii,:] = run_circuit(qc=qc,shots=shots,normalization_constant=normalization_constant, recreate_paper_results=True)
```
