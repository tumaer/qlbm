```python
import numpy as np
from QLBM_functions import create_QLBM_circuit, run_circuit

import matplotlib.pyplot as plt
```

```python
N_points_x = 32
x = np.linspace(0, 1, N_points_x)
advection_velocity = 0.1 * x + 0.1
timesteps = 50

shots = 10**7
initial_density = np.ones(N_points_x)*0.1
qc, normalization_constant = create_QLBM_circuit(initial_density, advection_velocity, timesteps)

density_result = run_circuit(qc=qc,shots=shots,normalization_constant=normalization_constant, recreate_paper_results=True)
```

```python
plt.plot(density_result)
plt.grid()
plt.xlim([0,N_points_x-1])
plt.xticks(np.arange(0, N_points_x, step=4))
plt.yticks(np.arange(0.06, 0.21, step=0.02))
plt.ylim([0.06,0.2])
plt.ylabel(r'$\rho$')
plt.xlabel(r'x')
plt.tight_layout()
plt.draw()
```
